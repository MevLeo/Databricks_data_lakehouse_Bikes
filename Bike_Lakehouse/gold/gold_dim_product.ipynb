#Transformation query

```python
query = """
select 
    row_number() over (order by cp.start_date, cp.product_Number) as product_key
    , cp.product_id
    , cp.product_Number
    , cp.product_name
    , cp.category_id
    , ec.category
    , ec.subcategory
    , ec.maintenance_flag
    , cp.product_line
    , cp.start_date
from workspace.silver.crm_products cp
left join workspace.silver.erp_product_category ec on cp.category_id = ec.category_id
   
"""

df = spark.sql(query)
```

```python
df.limit(10).display()
```

#witre from dataframe to gold table

```python
%sql
drop table if exists workspace.gold.dim_products
```

```python
df.write.mode("overwrite").format("delta").saveAsTable("workspace.gold.dim_products")
```
