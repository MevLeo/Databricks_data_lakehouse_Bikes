#Transformation query

```python
query = """
select 
    cs.order_number
    , dp.product_key
    , dc.customer_key
    , cs.order_date
    , cs.ship_date
    , cs.due_date
    , cs.sales_amount
    , cs.quantity
    , cs.price
from workspace.silver.crm_sales cs
left join workspace.gold.dim_customers dc on cs.customer_id = dc.customer_id
left join workspace.gold.dim_products dp on cs.product_number = dp.product_Number;
"""

df= spark.sql(query)
```

```python
df.limit(10).display()
```

#writing from dataframe to gold table 

```python
df.write.mode("overwrite").format("delta").saveAsTable("workspace.gold.fact_sales")
```
